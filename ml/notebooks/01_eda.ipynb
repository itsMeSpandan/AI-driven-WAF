# AI-Driven WAF — 01: Exploratory Data Analysis (EDA)

**Objective**: Analyze class distributions, request length statistics, duplicate rates, and verify dataset split integrity according to `ml/SCHEMA.md`.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Locate sample dataset
sample_csv_path = Path('../data/sample/requests_sample.csv')
print(f"Loading sample data from: {sample_csv_path.resolve()}")
df = pd.read_csv(sample_csv_path)
print(f"Dataset shape: {df.shape}")
df.head()
```

## 1. Class & Attack Type Balance

```python
print("Binary Label Distribution (0=Benign, 1=Threat):")
print(df['label'].value_counts(normalize=True).round(3))

print("\nAttack Taxonomy Distribution:")
print(df['attack_type'].value_counts())
```

## 2. Text Length Distributions
Character count and whitespace-delimited token count.

```python
df['char_length'] = df['text'].apply(len)
df['token_count'] = df['text'].apply(lambda x: len(x.split()))

df.groupby('attack_type')[['char_length', 'token_count']].describe().round(1)
```

## 3. Duplicate Detection & Split Integrity
Confirm no duplicate payloads exist across train, validation, test, and evasion splits (Zero Leakage Rule).

```python
duplicates = df.duplicated(subset=['text'], keep=False)
print(f"Total duplicated texts: {duplicates.sum()}")

print("\nPartition split breakdown:")
print(df['split'].value_counts())
```
